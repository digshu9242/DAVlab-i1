```python
# DECISION TREE CLASSIFIER - BREAST CANCER DATASET

# STEP 1: Import the required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

# STEP 2: Generate a matching synthetic SEER Breast Cancer dataset
np.random.seed(42)
n_samples = 1000

# Generate realistic correlations so the Decision Tree learns meaningful splits
tumor_sizes = np.random.randint(5, 100, size=n_samples)
nodes_positive = np.random.randint(1, 20, size=n_samples)
estrogen_status = np.random.choice(["Positive", "Negative"], p=[0.85, 0.15], size=n_samples)

# Make 'Status' depend slightly on tumor size, positive nodes, and estrogen status
risk_score = (tumor_sizes > 50).astype(int) + (nodes_positive > 8).astype(int) + (estrogen_status == "Negative").astype(int)
status_labels = np.where(
    risk_score >= 2,
    np.random.choice(["Alive", "Dead"], p=[0.4, 0.6], size=n_samples),
    np.random.choice(["Alive", "Dead"], p=[0.92, 0.08], size=n_samples)
)

cancer_df = pd.DataFrame({
    "Age": np.random.randint(30, 80, size=n_samples),
    "Race": np.random.choice(["White", "Black", "Other"], p=[0.8, 0.1, 0.1], size=n_samples),
    "Marital Status": np.random.choice(["Married", "Single", "Divorced", "Widowed"], size=n_samples),
    "T Stage": np.random.choice(["T1", "T2", "T3", "T4"], p=[0.4, 0.4, 0.15, 0.05], size=n_samples),
    "N Stage": np.random.choice(["N1", "N2", "N3"], p=[0.7, 0.2, 0.1], size=n_samples),
    "6th Stage": np.random.choice(["IIA", "IIB", "IIIA", "IIIB", "IIIC"], size=n_samples),
    "differentiate": np.random.choice(["Moderately differentiated", "Poorly differentiated", "Well differentiated"], size=n_samples),
    "Grade": np.random.choice(["1", "2", "3"], size=n_samples),
    "A Stage": np.random.choice(["Regional", "Distant"], p=[0.95, 0.05], size=n_samples),
    "Tumor Size": tumor_sizes,
    "Estrogen Status": estrogen_status,
    "Progesterone Status": np.random.choice(["Positive", "Negative"], p=[0.75, 0.25], size=n_samples),
    "Regional Node Examined": np.random.randint(1, 35, size=n_samples),
    "Reginol Node Positive": nodes_positive,
    "Survival Months": np.random.randint(1, 107, size=n_samples),
    "Status": status_labels
})

# Clean column names by removing extra spaces
cancer_df.columns = cancer_df.columns.str.strip()

print("First 5 rows of the dataset:")
print(cancer_df.head())
```
