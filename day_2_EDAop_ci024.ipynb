```python
import pandas as pd
import numpy as np

# loading the dataset first
data = pd.read_csv("diabetes_prediction_dataset.csv")
print(data.head())
```

```python
# checking last few rows also
print(data.tail())
```

```python
# basic size and column details
print("Shape:", data.shape)
print("Columns:")
print(data.columns)
```

```python
# checking what type of data is there
print(data.dtypes)
```

```python
# little more information about the dataframe
data.info()
```

```python
# some basic statistics
print(data.describe())
```

```python
# doing some statistics on age
print("Mean:", data["age"].mean())
print("Median:", data["age"].median())
print("Mode:")
print(data["age"].mode())
print("Variance:", data["age"].var())
print("Standard deviation:", data["age"].std())
```

```python
# minimum, maximum and number of age values
print("Minimum age:", data["age"].min())
print("Maximum age:", data["age"].max())
print("Count:", data["age"].count())
```

```python
# checking the smoking history values
print(data["smoking_history"].unique())
print("Number of unique values:", data["smoking_history"].nunique())
print(data["smoking_history"].value_counts())
```

```python
# checking missing values
print(data.isnull().sum())
```

```python
# percentage of missing values
missing_percent = data.isnull().mean() * 100
print(missing_percent)
```

```python
# making two versions just to see how missing values can be handled
clean_data = data.dropna()
filled_data = data.fillna(0)

print("Rows after dropna:", clean_data.shape[0])
print("Rows after fillna:", filled_data.shape[0])
```

```python
# checking duplicate rows
print("Duplicate rows:", data.duplicated().sum())

no_duplicates = data.drop_duplicates()
print("Rows after removing duplicates:", no_duplicates.shape[0])
```

```python
# selecting some data
print(data["age"].head())
print(data[["age", "bmi"]].head())
```

```python
# using iloc for first few rows
print(data.iloc[0:5])
```

```python
# filtering people whose age is above 30
above_30 = data[data["age"] > 30]
print(above_30.head())
print("Number of rows:", len(above_30))
```

```python
# sorting by age
sorted_data = data.sort_values("age")
print(sorted_data.head())
```

```python
# changing the age column name
changed_data = data.rename(columns={"age": "Age_Years"})
print(changed_data[["Age_Years", "bmi"]].head())
```

```python
# adding a small temporary column
changed_data["BMI_Adjustment"] = changed_data["bmi"] * 0.10
print(changed_data[["bmi", "BMI_Adjustment"]].head())

# removing the temporary column again
changed_data = changed_data.drop("BMI_Adjustment", axis=1)
```

```python
# grouping based on gender
gender_group = data.groupby("gender")
print(gender_group["bmi"].mean())
```

```python
# combining two small parts of the dataframe
part1 = data.iloc[:5]
part2 = data.iloc[5:10]

combined_data = pd.concat([part1, part2])
print(combined_data)
```

```python
# simple merge example
merge_data = data.copy()
merge_data["ID"] = np.arange(len(merge_data))

left_part = merge_data.iloc[:5]
right_part = merge_data.iloc[:5][["ID", "bmi", "diabetes"]]

final_merge = pd.merge(left_part, right_part, on="ID")
print(final_merge)
```
