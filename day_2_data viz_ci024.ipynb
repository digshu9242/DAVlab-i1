```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# first load the iris file
data = pd.read_csv("Iris.csv")
print(data)
```

```python
# checking the column names once
print(data.columns)
```

```python
# seeing how sepal length values are spread
sns.histplot(data=data, x="SepalLengthCm", bins=10)
plt.title("Sepal Length")
plt.xlabel("Sepal Length")
plt.ylabel("Count")
plt.show()
```

```python
# simple scatter plot for petal values
plt.scatter(data["PetalLengthCm"], data["PetalWidthCm"])
plt.title("Petal Length and Petal Width")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.show()
```

```python
# checking the petal width using box plot
plt.boxplot(data["PetalWidthCm"])
plt.title("Petal Width Box Plot")
plt.show()
```

```python
# now checking sepal length and width
plt.scatter(data["SepalLengthCm"], data["SepalWidthCm"])
plt.title("Sepal Length vs Sepal Width")
plt.xlabel("Sepal Length")
plt.ylabel("Sepal Width")
plt.show()
```

```python
# different species are shown separately here
sns.scatterplot(data=data, x="PetalLengthCm", y="PetalWidthCm", hue="Species")
plt.title("Petal values for different species")
plt.xlabel("Petal Length")
plt.ylabel("Petal Width")
plt.show()
```

```python
# pair plot gives many plots together
clean_data = data.drop(columns=["Id"])
sns.pairplot(clean_data, hue="Species")
plt.show()
```

```python
# checking how the numerical values are related
num_data = data.drop(columns=["Id"]).select_dtypes("number")
corr = num_data.corr()
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, fmt=".2f", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()
```
