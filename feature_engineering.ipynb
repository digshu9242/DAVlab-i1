```python
# Feature Engineering: Normalization, Standardization and PCA

# Step 1: Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.decomposition import PCA

# Step 2: Create a sample dataset
student_records = {
    'Study_Hours': [2, 4, 6, 8, 10, 3, 5, 7, 9, 12],
    'Attendance': [60, 65, 70, 75, 80, 62, 68, 78, 85, 95],
    'Assignment_Score': [40, 50, 55, 65, 70, 45, 60, 72, 80, 90],
    'Exam_Score': [45, 50, 60, 65, 75, 48, 58, 70, 82, 95]
}

raw_df = pd.DataFrame(student_records)

# Display original dataset
print("=" * 60)
print("ORIGINAL DATASET")
print("=" * 60)
print(raw_df.to_string(index=False))

print("\nSTATISTICAL SUMMARY")
print(raw_df.describe().round(2))

# Step 3: Normalization using MinMaxScaler
print("\n" + "=" * 60)
print("NORMALIZATION USING MINMAXSCALER")
print("=" * 60)

mm_scaler = MinMaxScaler()
scaled_minmax_arr = mm_scaler.fit_transform(raw_df)
minmax_df = pd.DataFrame(
    scaled_minmax_arr,
    columns=raw_df.columns
)

print("\nNormalized Dataset:")
print(minmax_df.round(2).to_string(index=False))

print("\nMinimum values:")
print(minmax_df.min().round(2))

print("\nMaximum values:")
print(minmax_df.max().round(2))

# Visualize normalized data
minmax_df.plot(
    kind='bar',
    figsize=(12, 5),
    title='Normalized Data (Min-Max Scaling)'
)
plt.xlabel("Student Index")
plt.ylabel("Normalized Value")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Step 4: Standardization using StandardScaler
print("\n" + "=" * 60)
print("STANDARDIZATION USING STANDARDSCALER")
print("=" * 60)

z_scaler = StandardScaler()
scaled_z_arr = z_scaler.fit_transform(raw_df)
zscore_df = pd.DataFrame(
    scaled_z_arr,
    columns=raw_df.columns
)

print("\nStandardized Dataset:")
print(zscore_df.round(2).to_string(index=False))

print("\nMean of each feature:")
print(zscore_df.mean().round(2))

print("\nStandard deviation of each feature:")
print(zscore_df.std(ddof=0).round(2))

# Visualize standardized data
zscore_df.plot(
    kind='bar',
    figsize=(12, 5),
    title='Standardized Data (Z-score Scaling)'
)
plt.xlabel("Student Index")
plt.ylabel("Standardized Value")
plt.axhline(0, color='black', linewidth=0.8)
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()

# Step 5: Apply PCA for dimensionality reduction
print("\n" + "=" * 60)
print("PRINCIPAL COMPONENT ANALYSIS (PCA)")
print("=" * 60)

# PCA should be applied after standardizing the features
pca_model = PCA(n_components=2)
reduced_components = pca_model.fit_transform(scaled_z_arr)
components_df = pd.DataFrame(
    reduced_components,
    columns=['PC1', 'PC2']
)

print("\nPCA Transformed Data:")
print(components_df.round(2).to_string(index=False))

# Step 6: Display explained variance
print("\nExplained Variance Ratio:")
print(np.round(pca_model.explained_variance_ratio_, 4))

print("\nExplained Variance Percentage:")
print(
    np.round(pca_model.explained_variance_ratio_ * 100, 2),
    "%"
)

retained_var = pca_model.explained_variance_ratio_.sum()
print("\nTotal Variance Retained:")
print(round(retained_var, 4))
print("Total Variance Retained (%):",
      round(retained_var * 100, 2), "%")

# Step 7: Visualize PCA output
plt.figure(figsize=(8, 5))
plt.scatter(
    components_df['PC1'],
    components_df['PC2'],
    color='blue',
    s=60
)

# Label each data point
for idx in range(len(components_df)):
    plt.annotate(
        str(idx + 1),
        (components_df['PC1'].iloc[idx], components_df['PC2'].iloc[idx]),
        xytext=(5, 5),
        textcoords='offset points'
    )

plt.xlabel("Principal Component 1 (PC1)")
plt.ylabel("Principal Component 2 (PC2)")
plt.title("PCA: Dimensionality Reduction")
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Step 8: Final comparison
print("\n" + "=" * 60)
print("DIMENSIONALITY REDUCTION SUMMARY")
print("=" * 60)
print("Original number of features:", raw_df.shape[1])
print("Number of PCA components:", components_df.shape[1])
print(
    "Variance retained by 2 components:",
    round(retained_var * 100, 2), "%"
)
print("\nProgram executed successfully!")
```
